### Structured output from LLM

```python
from agents import Runner, Agent , function_tool
from pydantic import BaseModel 
from dotenv import load_dotenv 


load_dotenv()
```

```python
class Recipe(BaseModel):
    name: str
    ingredients: list[str]
    instructions: str
    cooking_time_minutes: int

recipe_agent = Agent(
    name = "Recipe Agent",
    instructions="An Agent that generates recipies based on user input",
    model="gpt-4o-mini",
    output_type=Recipe
)    


user_input = "I have paneer, rice, and broccoli. What can I make for dinner?"
response = await Runner.run(recipe_agent, user_input)
print(response.final_output)
```

### Tool Calling 

```python
@function_tool
def get_weather(city: str) -> str:
    """Get the current weather for a given city."""
    print("get weather tool calling")
    return "sunny"

@function_tool
def get_temperature(city: str) -> str:
    """Get the current temperature for a given city."""
    print("get temperature tool calling")
    return "25 degrees Celsius"


weather_agent = Agent(
    name="Weather Agent",
    instructions="An Agent that provides weather information based on user input",
    model="gpt-4o-mini",
    tools=[get_weather, get_temperature]
)

response = await Runner.run(weather_agent, "what is the current weather and temperature of kolkata")
print(response.final_output)
```
