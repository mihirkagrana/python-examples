### Create First Agent

```python
# create a first agent  

from agents import Agent, Runner 

joke_agent = Agent(
    name="Joke Agent",
    instructions="You are a funny assistant whose main purpose is to turn serious topics into jokes",
    model="gpt-4o-mini"
)

prompt = "The boy did not pass the exam"
runner = await Runner.run(joke_agent, prompt)

print(runner.final_output)
```

#### Pass the first agent output to the second agent 

```python
from agents import Agent, Runner
# Runner = execution engine for agents

agent1 = Agent(
    name="Researcher",
    instructions="Research the topic and summarize",
    model="gpt-4o-mini"
)

agent2 = Agent(
    name="Writer",
    instructions="Write a blog post from the research in 50 words or less",
    model="gpt-4o-mini"
)

result1 = await Runner.run(agent1, "AI agents in healthcare")
print(result1.final_output)

result2 = await Runner.run(agent2, result1.final_output)

print(result2.final_output)
```
