### Using Ollama Library

```python
import ollama
```

```python
response = ollama.generate(model="gemma3:4b", prompt="What is the capital of France?")
print(response.response)
```

```python
response = ollama.generate(model="gemma3:4b", prompt="What is the capital of France?", stream=True)
for chunk in response:
    print(chunk.response, end="")
```

#### Pass image in the model

```python
import base64 

encoded_string = ''
with open('cat.png', 'rb') as f: 
    encoded_string = base64.b64encode(f.read()).decode('utf-8')

response = ollama.generate(model="gemma3:4b", images=[encoded_string], prompt="what is the color of the cat in the image", stream=True) 
for chunk in response: 
    print(chunk.response, end="")
    
       
```

#### Chat Message 

Generate histrory of the conversation 

```python
message_history = []

schema = {
    "type": "object",
    "properties": {
        "answer": {"type": "string"},
        "confidence": {"type": "number"}
    },
    "required": ["answer", "confidence"]
}

while True: 
    user_input = input("User: ")
    message_history.append({"role": "user", "content": user_input})

    response = ollama.chat(model="gemma3:4b", messages=message_history, stream=True, format=schema)
    
    assistant_message = ""
    for chunk in response: 
        assistant_message += chunk.message.content 
        print(chunk.message.content, end="")

    message_history.append({"role": "assistant", "content": assistant_message})    
```
